# HL7-to-FHIR Patient Data Mapping

Beginner Python project for simulating a simple EMR-to-data-exchange workflow.

### Project objective

Extract basic patient information from a simplified HL7-style message and map it to a simplified FHIR Patient resource.

**Note:** This is an educational prototype, not a production HL7/FHIR implementation.

```python
# 1. Sample simplified HL7-style patient message

hl7_message = (
    "MSH|^~\\\\&|Hospital|EMR|FHIR|System|20260929||ADT^A01|1001|P|2.5\n"
    "PID|1||P1001||Apoorva^Dogra||19991104|F"
)

print(hl7_message)
```

```python
# 2. Split the message into segments

segments = hl7_message.split("\n")

patient_segment = None

for segment in segments:
    if segment.startswith("PID|"):
        patient_segment = segment
        break

print(patient_segment)
```

```python
# 3. Extract patient information from the PID segment

if patient_segment:
    fields = patient_segment.split("|")

    patient_id = fields[3]
    name = fields[5].split("^")
    date_of_birth = fields[7]
    gender = fields[8]

    first_name = name[0]
    last_name = name[1]

    print("Patient information extracted from HL7 message:")
    print("Patient ID:", patient_id)
    print("First Name:", first_name)
    print("Last Name:", last_name)
    print("Date of Birth:", date_of_birth)
    print("Gender:", gender)
else:
    print("No patient information found.")
```

```python
# 4. Map the information to a simplified FHIR Patient model

fhir_patient = {
    "resource_type": "Patient",
    "id": patient_id,
    "name": {
        "given": first_name,
        "family": last_name
    },
    "birth_date": date_of_birth,
    "gender": gender
}

fhir_patient
```

```python
# 5. Display the mapped FHIR-style resource

print("FHIR-style Patient resource:")
print("Resource Type:", fhir_patient["resource_type"])
print("Patient ID:", fhir_patient["id"])
print("Given Name:", fhir_patient["name"]["given"])
print("Family Name:", fhir_patient["name"]["family"])
print("Birth Date:", fhir_patient["birth_date"])
print("Gender:", fhir_patient["gender"])
```

### Mapping summary

The simplified HL7 PID fields have been extracted and mapped to corresponding FHIR Patient fields.
